### Logical Operations
- AND --> All the conditions must be True
- OR --> Alteast one condition must be True
- Not --> It reverse the conditions

```python
### AND,OR,NOT


#AND
x=True
y=True
r=x and y
print(r)
```

```python
#OR
x=True
y=False
r=x or y
print(r)
```

```python
##NOT
x=True
r=not(x)
print(r)
```
