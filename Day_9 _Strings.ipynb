### Strings
- It is a sequence of characters enclosed with quotes -"",'',''' '''.
- Strings can store alphabets, Numbers and even special characters 


```python
### Create A String 
Name="Vivek"
greeting ="Hello Everyone"
mutiline=''' THis a  mutline 
string'''
```

```python
### Accessing The String Elements
word="Python"
print(word[0])
print(word[-1])
print(word[1])
```

```python
## String Slicing  
word="PythonProgramming"
print(word[0:6])
print(word[6:])
print(word[:6])
print(word[0:10:3])
```

```python
## String Immutability
## strings in python are immutable, meaning you cannot modify then directly 
word="Hello"
word[0]="J"
```

```python
word="j"+word[1:]
print(word)
```

```python
## String Methods
name="Sai Vivek"
print(name.upper()) ## Upper case
print(name.lower()) ## Lower Case
print(name.title()) ## Starting Letter  capital for each word
print(name.capitalize()) ## Staring letter Capital
print(name.swapcase())  ## Convert lower to upper and viceversa
```

```python
### String Type
text="abc"
print(text.isalpha()) ## checking the string contains alphabets
print(text.isdigit()) ## checking the string contains digits
print(text.isalnum()) ## checking the sting contains either number or alphabets
print(text.isspace()) ##  checking the string contains a space
```

```python
## Search and Replace
sentence="I Like Dhoni"
print(sentence.find("Like"))
print(sentence.replace("Like","Love"))
```

```python
### Splitting and Joininig 

words="Hello World Python".split()
print(words)

joined="-".join(words)
print(joined)
```

```python
## Strip Methods
data="  Devops "
print(data.strip()) ## remove extra spaces
print(data.lstrip())  ## remove left side extra spaces
print(data.rstrip()) ## remove right side extra spaces
```

```python
### String Formatting
name="viv"
age=22

## Old Method
print("My Name is %s and I am %d Years Old."%(name,age))

## Format() Method
print("My Name is {} and I am {} Years Old.".format(name,age))

## Best and Modern Method ->F-strings
print(f"My Name is {name} and I am {age} Years Old.")
```

```python
## String Comparsion and Membership

a="hello"
b="world"
print(a==b)
print(a<b)  ## Lexicographical order
print("h" in a)
print("z" in b)
```
