### LOOPS
- A Loop is a programming structure that repeatedly exectues a block of code
- A loop allows a program to perform the same task multiple times automatically

### Types of Loops
- For loop
- While loop
- loop control Statements
- Nested Loops

  


```python
# Print 1st 5 numbers
# for loop

for i  in range(6):
    print(i,end=" ")
```

```python
for i in range(1,6):
    print(i,end=" ")
```

```python
for i  in range(1,10):
    print(i,end=" ")
```

```python
for i  in range(10,1,-1):
    print(i,end=" ")
```

```python
### Strings

name="Viv God"
for i in name:
    print(i)
    
```

```python
### While Loop 
### The wwhile loop continues to execute as long as the condition is True

c=0
while c<5:
    print(c)
    c=c+1
```

```python
### Loop Control Statements

## Break 
## The Break Statement exits the loop prematurely

for i in range(10):
    if i==5:
        break
    print(i)
    
    
```

```python
### Continue
## The Continue Statement skips the current iteration and continues with next

for i in range(10):
    if i %2==0:
        continue
    print(i)
```

```python
### Pass
### The Pass Statement is  a null operation; It does nothing.

for i in range(5):
    if i==3:
        pass ## placeholder
    print(i,end=" ")
```

```python
### print 5 table using loops
n=int(input("Enter the number"))
for i in range(1,11):
    print(n,"X",i,"=",n*i)
```

```python
### Nested Loops
### A Loop inside a Loop

for i in range(3): ## outerloop
    for j in range(3): ## innerloop
        print(f"i:{i} and j:{j}")
```

```python
### Print the multiplication Tables from 1 to 10

for i in range(1,11):
    print("Table of ",i)
    for j in range(1,11):
        print(i,"X",j,"=",i*j)
    print("____________________________") 
        
```

### Tasks of the  Day
1.Calculate the sum of first n natural numbers using  for and while loop
2.Print prime numbers between 1 and 100 using for loop


```python
n=int(input("enter the number"))
c=0
for i in range(1,n+1):
    c=c+i
print("sum of n natural number",c)
    
```

```python
n = int(input("Enter n: "))

sum = 0
i = 1

while i <= n:
    sum = sum + i
    i += 1

print("Sum of first", n, "natural numbers is:", sum)
```

```python
for num in range(2, 101):
    count = 0

    for i in range(1, num + 1):
        if num % i == 0:
            count += 1

    if count == 2:
        print(num, end=" ")
```
