### File Handling 
- File Handling is A  Crucial part of any programmimg language
- It provides built-in functions and methods to read from write to files.

```python
### How to Read A File

with open('vivek.txt','r') as file:
    content=file.read()
    print(content)
```

```python
### Read a file line by line
with open('vivek.txt','r') as file:
    for line in file:
        print(line.strip()) ## strip() removes the newline characters
```

```python
### writing a File
with open('vivek.txt','w') as file:
    file.write("Hello world!\n")
    file.write("This is a new line!\n")
    file.write("This is Third Line!!!")
    
```

```python
### Write a FIle (without overriding)
with open('vivek.txt','a') as file:
    file.write("appending this line!!\n")
```

```python
## Read the content from a source tect file and  write to a destination text file

with  open('vivek.txt','r') as source_file:
    content=source_file.read()
with open('destination.txt','w') as destination_file:
    destination_file.write(content)
    
```

```python
### Read a text file and count the number of lines, words and characterd
### counting lines,words and characters in a text file
with open('vivek.txt', 'r') as file:
    line_count = 0
    word_count = 0
    char_count = 0

    for line in file:
        line_count += 1
        word_count += len(line.split())
        char_count += len(line)

print("Number of lines:", line_count)
print("Number of words:", word_count)
print("Number of characters:", char_count)
```

```python
def count_text_file(file_path):
    with open(file_path,'r') as file:
        lines=file.readlines()
        line_count=len(lines)
        word_count=sum(len(line.split()) for line in lines)
        char_count=sum(len(line) for line in lines)
    return line_count,word_count,char_count
file_path='vivek.txt'
lines,words,characters=count_text_file(file_path)
print(f"line:{lines},words:{words},char:{characters}")
```

#### W+ mode
- The w+ mode in python is used to open file for both reading and writing
- If the file doesnt exist it will be created
- if the files exists, the content will be  truncated(Override)

```python
### write and then read the file
with open('vivek.txt','w+') as file:
    file.write("welcome to vivgod arena\n")
    file.write("Hello vivgod\n")

    ### move the file cursor to the begining
    file.seek(0)

    ## read the content of the file
    content=file.read()
    print(content)
    
```

### Handling File Paths
- While working with files in python , Handling file paths correctly is crucial
- This will ensure your code works  across different operating systems and enviornments
  

```python
import os
os.getcwd()
```

```python
## check if a path is a file or directory
import os
path='vivek.txt'
if os.path.isfile(path):
    print(f"The path {path} is a file")
elif os.path.isdir(path):
    print(f"The path {path} is a directory")
else:
    print(f"The path {path} is not a file or directory")
```
