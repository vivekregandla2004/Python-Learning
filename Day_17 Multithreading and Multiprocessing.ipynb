### Multithreading and Multiprocessing

-  Multithreading: Running Multiple Threads parallely in a process is called Multithreading

```python
import time
def task1():
    print("Task1 Started.........")
    time.sleep(2)
    print("Task1 Ended...........")
def task2():
    print("Task2 Started.........")
    time.sleep(3)
    print("Task2 Ended...........")
start=time.time()
task1()
task2()
end=time.time()
print(f"Task is done in {end-start}")
```

```python
import time
import threading
def task1():
    print("Task1 Started.........")
    time.sleep(2)
    print("Task1 Ended...........")
def task2():
    print("Task2 Started.........")
    time.sleep(3)
    print("Task2 Ended...........")

t1=threading.Thread(target=task1)
t2=threading.Thread(target=task2)
start=time.time()
t1.start()
t2.start()
t1.join()
t2.join()
end=time.time()
print(f"Task is done in {end-start}")
```

```python
import threading
import requests
import time

def fetch_data(api_name,url):
    print(f"API NAME:{api_name}")
    data=requests.get(url)
    time.sleep(2)
    print("Data Fetching.....")
    res=data.json()
    print(f"Length of Data:{len(res)}")
t1=threading.Thread(target=fetch_data,args=("Products","https://fakestoreapi.com/products"))
t2=threading.Thread(target=fetch_data,args=("Typicdoe","https://jsonplaceholder.typicode.com/todos"))

start=time.time()
t1.start()
t2.start()
t1.join()
t2.join()
end=time.time()
print(f"Task is done in {end-start}")
```

### GIL
- Python Global Interpreter Lock (GIL) is a mutex (mutual-exclusion lock) in the default CPython interpreter that restricts the execution of Python bytecode to exactly one thread at a time, even on multi-core processors

## MultiProcessing

```python
from multiprocessing import Process
import time

def waiter():
    print("Taking the order")
    time.sleep(2)
    print("Done")
def chef():
    print("Cooking Food")
    time.sleep(5)
    print("Done")
def server():
    print("Food is servered")
    time.sleep(2)
    print("Done")
p1=Process(target=waiter())
p2=Process(target=chef())
p3=Process(target=server())

start=time.time()
p1.start()
p2.start()
p3.start()

p1.join()
p2.join()
p3.join()
end=time.time()
print(f"Task is done in {end-start}")


    
```
