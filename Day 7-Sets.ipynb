#### Sets
- Sets are a built in data type in python  to store  the collection of unique items
- They are unordered --> they don't follow a specific order
- They do not allow duplicates
- sets are useful for membership tests, eliminating duplicate entires and performing mathematical set operations

```python
### Create a Set
my_set={1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set=set()
print(type(my_empty_set))
```

```python
my_set=set([1,2,3,4,5,6])
print(my_set)
```

```python
### Basic Set operations 
my_set.add(7)
print(my_set)
my_set.add(8)
print(my_set)
```

```python
### Remove the elements from a set
my_set.remove(3)
print(my_set)
```

```python
my_set.remove(10)
```

```python
my_set.discard(7)
print(my_set)
```

```python
my_set.discard(11)
print(my_set)
```

```python
### pop method
removed_element=my_set.pop()
print(removed_element)
print(my_set)  ## in set the pop method is works like the they pop in different order 
```

```python
### Set Membership Test
my_set={1,2,3,4,5}
print(3 in my_set)
print(10 in my_set)
```

```python
### Mathematical Operations
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

## Union
union_set=set1.union(set2)
print(union_set)

### intersection
intersection_set=set1.intersection(set2)
print(intersection_set)

###Intersectiion_Update
set1.intersection_update(set2)
print(set1)

###
```

```python
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}

### Difference
print(set1.difference(set2))
print(set2.difference(set1))
```

```python
### Symmetric Difference
set1.symmetric_difference(set2)
```

```python
### Set Methods
set1={1,2,3,4,5}
set2={3,4,5}

## subset what are the elements in the set 1 that all values  need to be present in set2
print(set1.issubset(set2))

## Superset what are the values i have in the set1 should be in set 2
print(set1.issuperset(set2))
```

```python
print(set2.issubset(set1))
print(set2.issuperset(set1))
```
