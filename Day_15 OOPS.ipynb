**Create a Class Atm which does the following**

1.Deposit
2.withdraw
3.Show_Balance

```python
class ATM:
    bank_name = "ABC Bank"

    def __init__(self):
        self.account = int(input("Enter Account Number: "))
        self.str_account = str(self.account)
        self.balance = 500

    def deposit(self):
        amount = int(input("Enter Amount to Deposit: "))
        self.balance += amount
        print(f"Your Account with Account Number XXXXXX{self.str_account[-4:]} has been credited with an amount of Rs {amount}")

    def withdraw(self):
        amount = int(input("Enter Amount to Withdraw: "))
        if amount % 100 == 0 and amount <= self.balance:
            self.balance -= amount
            print(f"Your Account with Account Number XXXXXX{self.str_account[-4:]} has been debited with an amount of Rs {amount}")
        else:
            raise Exception("Insufficient Balance or Amount is not a multiple of 100.")

    def show_balance(self):
        print(f"Current Balance of Account XXXXXX{self.str_account[-4:]} is Rs {self.balance}")


# Creating object
a1 = ATM()

# Testing
a1.deposit()
a1.withdraw()
a1.show_balance()
```

### Library Management  System
- Add Book
- issue Book
- return book
- must return book in 7 days he need to pay a fine 10rupees


```python
class Library:
    fine = 10

    def __init__(self):
        self.books = []
        self.issued_books = {}

    def add_book(self):
        book = input("Enter book name: ")
        self.books.append(book)
        print(f"{book} added successfully.")

    def issue_book(self):
        book = input("Enter book name to issue: ")

        if book in self.books:
            name = input("Enter student name: ")
            day = int(input("Enter issue day (1-31): "))

            self.books.remove(book)
            self.issued_books[book] = [name, day]

            print(f"{book} issued to {name}.")
        else:
            print("Book is not available.")

    def return_book(self):
        book = input("Enter book name to return: ")

        if book in self.issued_books:
            return_day = int(input("Enter return day (1-31): "))

            issue_day = self.issued_books[book][1]
            days = return_day - issue_day

            if days > 7:
                fine = (days - 7) * Library.fine
                print(f"Late return. Fine = Rs {fine}")
            else:
                print("Book returned on time. No fine.")

            self.books.append(book)
            del self.issued_books[book]

        else:
            print("This book was not issued.")

    def show_books(self):
        print("Available Books:")
        for book in self.books:
            print(book)
l1 = Library()

while True:
    print("\n===== Library Management System =====")
    print("1. Add Book")
    print("2. Issue Book")
    print("3. Return Book")
    print("4. Show Available Books")
    print("5. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        l1.add_book()

    elif choice == 2:
        l1.issue_book()

    elif choice == 3:
        l1.return_book()

    elif choice == 4:
        l1.show_books()

    elif choice == 5:
        print("Thank You!")
        break

    else:
        print("Invalid Choice")
```

## 3.Abstraction 
- the process of Hiding  the unneccessary implementation and showing only the necessary things .
- We are Creating Rule
      : Withdraw is a mandatory rule.
To Acheive Abstraction we require two things:
1. we import ABC class from abc
       - ABC-Abstract Base Class
2. At least one abstract method must be there in Abstract class(Blue print class).


```python
## Abstraction
from abc import ABC,abstractmethod

class ATM(ABC):
    @abstractmethod
    def withdraw(self):
        pass
```

```python
atm=ATM()
```

```python
class SBI(ATM):
    def withdraw(self):
        self.__check_balance()
        self.__check_pin()
        print("WithDrawing...")
        print("Withdraw successfull!!!")
    def __check_balance(self):
        print("Checking Balance")
    def __check_pin(self):
        print("Verifying Pin")
```

## Restaurant Management System(RMS)
1. Order food()
2. bill()
3. list_items()
4. add_items()
5. book table()
   

```python
class Restaurant:

    def __init__(self):
        self.menu = {
            "Pizza": 250,
            "Burger": 150,
            "Biryani": 300
        }

    def list_items(self):
        print("\n------ MENU ------")
        for item, price in self.menu.items():
            print(item, ":", price)

    def add_items(self):
        item = input("Enter item name: ")
        price = int(input("Enter price: "))
        self.menu[item] = price
        print(item, "added successfully!")

    def order_food(self):
        item = input("Enter item name: ")
        quantity = int(input("Enter quantity: "))

        if item in self.menu:
            total = self.menu[item] * quantity
            print("\nOrder Placed Successfully!")
            print("Item:", item)
            print("Quantity:", quantity)
            print("Total Bill: ₹", total)
        else:
            print("Item not available!")

    def bill(self):
        item = input("Enter item name: ")
        quantity = int(input("Enter quantity: "))

        if item in self.menu:
            print("Bill Amount: ₹", self.menu[item] * quantity)
        else:
            print("Item not found!")

    def book_table(self):
        name = input("Enter your name: ")
        table = int(input("Enter table number: "))
        print(f"Table {table} booked successfully for {name}.")


# Object Creation
r = Restaurant()

while True:
    print("\n===== RESTAURANT MANAGEMENT SYSTEM =====")
    print("1. List Items")
    print("2. Add Item")
    print("3. Order Food")
    print("4. Bill")
    print("5. Book Table")
    print("6. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        r.list_items()

    elif choice == 2:
        r.add_items()

    elif choice == 3:
        r.order_food()

    elif choice == 4:
        r.bill()

    elif choice == 5:
        r.book_table()

    elif choice == 6:
        print("Thank you! Visit Again 😊")
        break

    else:
        print("Invalid Choice!")
```
