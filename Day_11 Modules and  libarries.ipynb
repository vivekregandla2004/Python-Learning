### What is a module 
- A python module is simply a python file
- One .py file = One Module

```python
import math
math.sqrt(16)
```

```python
from math import  sqrt
sqrt(256)
```

### Why import is important
- By importing a module, You are importing thousands of lines of Code already tested ,optimized, debugged by experts
- You Save time
- Your code becomes smaller
- Your code becomes reliable
- Your code is easier to maintain

```python
import math  as m
m.sqrt(25)
```

### User Defined Modules
- Modules created by Us

```python
import greetings
greetings.hello("vivek")
greetings.bye("god")
```

### What is a Package?
Package  is a collections of modules stored in a folder
- Module --> A single .py file
- Package --> Folder containing multiple modules

```python
from mypackage import greet
greet.hello("vivek")
```

```python
import mypackage.greet as g
g.bye("god")
```

```python
import mypackage.maths as m
m.add(2,5)
```

```python
import mypackage.maths as m
m.sub(12,5)
```

### Libraries
Python comes with a toolbox of pre-written modules called **Standard Libraries**
- A group of Packages are nothing  but Library

```python
import random
print(random.randint(1,10))
print(random.choice(['apple','banana','cherry']))
```
