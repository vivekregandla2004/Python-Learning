## Generators

- A Python generator is a special type of function that returns an iterator object, allowing you to produce a sequence of values on the fly (lazy evalution) instead of computing them all at once and saving them to memory

```python
l1=[1,2,3,4,5,6]
```

```python
def display():
    yield 1
    yield 2
    yield 3
```

```python
x=display()
```

```python
print(type(x))
```

```python
for i in x:
    print(i)
```

```python
print(next(x))
```

```python
def display():
    return 1
    return 2
    return 3
```

```python
x=display()
```

```python
print(x)
```

```python
print(type(x))
```

## Decorators
- A Python Decorator is a deisgn pattern that allows you to modify or extend the behavior of a function or method without changing its actual source code.

```python
def display():
    print("Welcome to Decorators")
```

```python
display()
```

```python
###  Creating Decorator Function
@decorator_func
def greet():
    print("Good Morning")
```

```python
def decorator_func(func):
    def wrapper():
        print("######################")
        func()
        print("######################")
    return wrapper
        
```

```python
##result=decorator_func(greet)
```

```python
##result()
```

```python
greet() ## direclty calling the function using the decorator
```

```python
is_loggedin=input("Enter Yes/No")
def login_required(func):
    def wrapper():
        if is_loggedin.lower() =="yes":
            func()
        else:
            print("please login first")
    return wrapper
@login_required
def dashboard():
    print("welcome to user dashboard")
dashboard()
```
