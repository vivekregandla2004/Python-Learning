#### Conditional Statements
- Conditional statements allow our program to make decisions
- In Programming, depending on conditions, we can make the program run different blocks of code

#### Different Conditional Statement in Python
- if
- Else
- Elif
- Nested Conditional Statements

```python
## IF Statement

age = 18
if age >= 18:
    print("You are allowed to Vote")
```

```python
## Else Statement
## The Else block executes when the condition in the IF statement is False

age = 20
if age >=18:
    print("You are eligible to Vote")
else:
    print("You are Minor")
```

```python
## ELif Statement
## Elif stands for "Else - if"

## This is useful when you have multiple conditions to check

age = 21
if age < 13:
    print("You are a Child")
elif age <= 19:
    print("You age a Tennages")
else:
    print("You are an Adult")
```

```python
## Nested Conditional Statements
## When we need to keep one condition inside another ---> Nesting

num  = int(input("Enter a Number: "))

if num > 0:
    print("The Number is Positive")
    if num % 2 == 0:
        print("The Number is Even")
    else:
        print("The Number is Odd")
else:
    print("The Number is either Zero or Negative")
```

```python
## Calculate the Leap Year 

year = int(input("Enter the Year:"))

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print(year, "is a leap year")
        else:
            print(year, "is not a leap year")
    else:
        print(year, "is a leap year")
else:
    print(year, "is not a leap year")
```
