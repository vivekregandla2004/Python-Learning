### Example 1: Create a todo list to keep track of items

```python
tasks = []

while True:
    print("\n TO DO LIST ")
    print("1. Add Task")
    print("2. Update Task")
    print("3. Delete Task")
    print("4. View Tasks")
    print("5. Exit")

    n = int(input("Enter your choice: "))

    if n == 1:
        task = input("Enter task: ")
        tasks.append(task)
        print("Task added successfully!")

    elif n == 2:
        if len(tasks) == 0:
            print("No tasks available.")
        else:
            print("Current Tasks:", tasks)
            index = int(input("Enter task number to update: ")) - 1

            if 0 <= index < len(tasks):
                tasks[index] = input("Enter updated task: ")
                print("Task updated!")
            else:
                print("Invalid task number")

    elif n == 3:
        if len(tasks) == 0:
            print("No tasks available.")
        else:
            print("Current Tasks:", tasks)
            index = int(input("Enter task number to delete: ")) - 1

            if 0 <= index < len(tasks):
                removed = tasks.pop(index)
                print(f"Deleted: {removed}")
            else:
                print("Invalid task number")

    elif n == 4:
        if len(tasks) == 0:
            print("No tasks found.")
        else:
            print("\nYour Tasks:")
            for i, task in enumerate(tasks, start=1):
                print(f"{i}. {task}")

    elif n == 5:
        print("Exiting Todo List...")
        break

    else:
        print("Invalid choice")
```

```python
to_do_list=["Buy Groceries","Clean the House","Pay the Bills"]

### Adding A Task

to_do_list.append("Go to GYM")
to_do_list.append("Walk  up early")

## removing A task

to_do_list.remove("Clean the House")

### Checking  if a task is in the list

if "Pay  the Bills" in to_do_list:
    print("Dont forget to pay the bills")
print("To do list remaining")
for task in to_do_list:
    print(f"-{task}")
```

```python
### Example 3
feedback=["great Service","very satisfied","could be better","excellent"]

### Adding a New Feedback
feedback.append("not happy with the service")

### COunting specific Feedback

postive_feedback_count=sum(1 for comment  in feedback if "great" in comment.lower() or "excellent" in comment.lower() or "satisfied" in comment.lower())
print(postive_feedback_count)

negative_feedback_count=sum(1 for comment  in feedback if "not" in comment.lower() or "could" in comment.lower())
print(negative_feedback_count)

print("User feedback:")
for c in feedback:
    print(f"-{c}")
```
