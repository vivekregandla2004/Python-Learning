### Exception Handling
- Exception Handling in Python allow us to handle errors without stopping the execution of our program
#### What are Exceptions
- Exceptions are events that disrupt the normal flow a program
- They occur when an error is encountered during program executions
    - ZeroDivisionError
    - FileNotFoundError
    - ValueError
    - TypeError

```python
#### Exception using try, except blocks
try:
    a=b
except:
    print("The  Variable has not been assigned")
```

```python
try:
    a=b
except NameError as x:
    print(x)
```

```python
try:
    result=1/0
except ZeroDivisionError as ex:
    print(ex)
    print("Please enter denominator greater than 0")
```

```python
try:
    num=int(input("enter a number"))
    result=10/num
except ValueError:
    print("This is not a Valid Number")
except ZeroDivisionError:
    print("Please Enter the number greater than 0")
except Exception as ex:
    print(ex)
else:
    print(f"result:{result}")
    
```

```python
### Try,except,Finally

try:
    num=int(input("enter a number"))
    result=10/num
except ValueError:
    print("This is not a Valid Number")
except ZeroDivisionError:
    print("Please Enter the number greater than 0")
except Exception as ex:
    print(ex)
else:
    print(f"result:{result}")
finally:
    print("Execution Completed")
```
